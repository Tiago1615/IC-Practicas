```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = pd.read_csv('SAMPLES_NORM.TXT', delim_whitespace=True, comment='#', names=['time', 'voltage'], header=None)

print("Datos cargados:")
print(data.head(10))
print(f"\nTotal de muestras: {len(data)}")
```

```python
plt.figure(figsize=(12, 6))
plt.plot(data['time'], data['voltage'], linewidth=2, color='blue', marker='o', markersize=3)
plt.xlabel('Tiempo (msecs)', fontsize=12)
plt.ylabel('Voltaje (mvolts)', fontsize=12)
plt.title('Señal Analógica - Tiempo vs Voltaje', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
